```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
from iminuit import Minuit
from iminuit import cost
from scipy import stats
from sympy import * 
import pandas as pd
r = np.random 
r.seed(42)
```

### 1.1.1

```python
r.seed(42)
t1=20
p1=0.5
t2=2500
p2=0.004
trials = 10**6
n1_bigger = 0
n2_bigger = 0
equals = 0
for i in range(trials):
    N1 = r.binomial(t1, p1)
    N2 = r.binomial(t2, p2)
    if N1 > N2:
        n1_bigger += 1
    if N2 > N1:
        n2_bigger += 1
    if N1 == N2:
        equals += 1
print(n1_bigger, n2_bigger, equals)
f1 = n1_bigger / trials
ef1 = np.sqrt(f1 * (1 - f1) / trials)
f2 = n2_bigger / trials
ef2 = np.sqrt(f2 * (1 - f2) / trials)
f3 = equals / trials
ef3 = np.sqrt(f3 * (1 - f3) / trials)
print(f'P(N_1> N_2)= {f1:.4f} ± {ef1:.4f}, P(N_2> N_1)= {f2:.4f} ± {ef2:.4f}, P(N_1=N_2)= {f3:.4f} ± {ef3:.4f}')
```

```python
N1 = stats.binom(n=20, p=0.5)
N2 = stats.binom(n=2500, p=0.004)

# Support ranges
k1 = np.arange(0, 21)
k2 = np.arange(0, 2501)

# PMFs
pmf1 = N1.pmf(k1)
pmf2 = N2.pmf(k2)

# Exact probabilities
P_N1_gt_N2 = sum(
    pmf1[i] * pmf2[:i].sum()
    for i in range(len(k1))
)

P_N2_gt_N1 = sum(
    pmf1[i] * pmf2[i+1:].sum()
    for i in range(len(k1))
)

P_equal = sum(
    pmf1[i] * pmf2[i]
    for i in range(len(k1))
)

print("Exact Binomial results:")
print(f"P(N1 > N2) = {P_N1_gt_N2:.4f}")
print(f"P(N2 > N1) = {P_N2_gt_N1:.4f}")
print(f"P(N1 = N2) = {P_equal:.4f}")
print(f"Check sum   = {P_N1_gt_N2 + P_N2_gt_N1 + P_equal:.4f}")
```

### 1.1.2

```python
r.seed(42)
t1=20
p1=0.5
t2=2500
p2=0.004
trials = 10**6
n1_hit = 0
n2_hit = 0
for i in range(trials):
    N1 = r.binomial(t1, p1)
    N2 = r.binomial(t2, p2)
    if 25 <= N1 <= 250:
        n1_hit += 1
    if 25 <= N2 <= 250:
        n2_hit += 1
print(n1_hit, n2_hit)
f1 = n1_hit / trials
ef1 = np.sqrt(f1 * (1 - f1) / trials)
f2 = n2_hit / trials
ef2 = np.sqrt(f2 * (1 - f2) / trials)
print('Simulated results')
print(f'P(25 <= N_1 <= 250)= {f1:.20f} ± {ef1:.20f}, P(25 <= N_2 <= 250)= {f2:.6f} ± {ef2:.6f}')
```

```python
f1_anal = stats.binom.sf(25, t1, p1) - stats.binom.sf(250, t1, p1)
f2_anal = stats.binom.sf(25, t2, p2) - stats.binom.sf(250, t2, p2)
print('analytical results:')
print(f'P(25 <= N_1 <= 250)= {f1_anal}, P(25 <= N_2 <= 250)= {(f2_anal):.6f}')
```

### 1.1.3

```python
r.seed(42)
t1=20
p1=0.5
t2=2500
p2=0.004
trials = 10**6
n1_10_hit = 0
n1_20_hit = 0
n2_10_hit = 0
n2_20_hit = 0
for i in range(trials):
    N1 = r.binomial(t1, p1)
    N2 = r.binomial(t2, p2)
    if N1 == 10:
        n1_10_hit += 1
    if N1 == 20:
        n1_20_hit += 1
    if N2 == 10:
        n2_10_hit += 1
    if N2 == 20:
        n2_20_hit += 1
print(n1_10_hit, n1_20_hit, n2_10_hit, n2_20_hit)
f1_10 = n1_10_hit / trials
ef1_10 = np.sqrt(f1_10 * (1 - f1_10) / trials)
f1_20 = n1_20_hit / trials
ef1_20 = np.sqrt(f1_20 * (1 - f1_20) / trials)
f2_10 = n2_10_hit / trials
ef2_10 = np.sqrt(f2_10 * (1 - f2_10) / trials)
f2_20 = n2_20_hit / trials
ef2_20 = np.sqrt(f2_20 * (1 - f2_20) / trials)
print('Simulated results')
print(f'P(N_1=10)= {f1_10:.4f} ± {ef1_10:.4f}, P(N_1=20)= {f1_20:.7f} ± {ef1_20:.7f}, P(N_2=10)= {f2_10:.4f} ± {ef2_10:.4f}, P(N_2=20)= {f2_20:.5f} ± {ef2_20:.5f}')
```

```python
f1_10_anal = stats.binom.pmf(10, t1, p1)
f2_10_anal = stats.binom.pmf(10, t2, p2)
print('analytical results:')
print(f'P(N_1=10)= {f1_10_anal:.5f}, P(N_2=10)= {f2_10_anal:.5f}')
f1_20_anal = stats.binom.pmf(20, t1, p1)
f2_20_anal = stats.binom.pmf(20, t2, p2)
print('analytical results:')
print(f'P(N_1=20)= {f1_20_anal:.7f}, P(N_2=20)= {f2_20_anal:.5f}')
```
